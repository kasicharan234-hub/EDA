```python
import pandas as pd
height = [120, 122, 125, 127, 121, 123, 137, 131, 161, 145, 141, 132]
bins = [118, 125, 135, 160, 200]
category = pd.cut(height, bins)
category
```

```python
pd.value_counts(category)
```

```python
bin_names = ['Short Height', 'Average height', 'Good Height', 'Taller']
pd.cut(height, bins, labels=bin_names)
```

```python
import pandas as pd
df = pd.read_csv('fraudTrain.csv')
print(df.head())
```

```python
df.head(10)
```

```python
amount = pd.to_numeric(df["amt"])
amount[np.abs(amount) > 100]
```

```python
CityPopulation = df["city_pop"]
df[CityPopulation > 2000]
```

```python
Q1 = df['amt'].quantile(0.25)
Q3 = df['amt'].quantile(0.75)
IQR = Q3 - Q1
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
print(f"Normal transaction range: ${lower_bound:.2f} to ${upper_bound:.2f}")
```

```python
df_clean = df[(df['amt'] >= lower_bound) & (df['amt'] <= upper_bound)]
df_clean
```
