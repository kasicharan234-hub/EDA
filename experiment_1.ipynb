```python
import pandas as pd
```

```python
df1Car =  pd.DataFrame({ 'VehicleID': [101, 103, 105, 107, 109, 111, 113, 115, 117, 119],
                       'parkingFeeC' : [40, 60, 50, 80, 30, 70, 90, 45, 55, 65]})
df2Car =  pd.DataFrame({'VehicleID': [102, 104, 106, 108, 110, 112, 114, 116, 118, 120],
                       'parkingFeeC': [ 35, 75, 50, 60, 45, 55, 65, 70, 80, 40]})
df1Bike =  pd.DataFrame({ 'VehicleID': [101, 103, 105, 107, 109, 111, 113, 115, 117,119],
                       'parkingFeeB' : [20, 25, 15, 30, 20, 35, 25, 30, 15, 20]})
df2Bike =  pd.DataFrame({'VehicleID': [102, 104, 106, 108, 110, 112, 114, 116, 118, 120],
                       'parkingFeeB': [15, 20, 25, 20, 30, 25, 35, 15, 20, 30]})
```

```python
dfCar = pd.concat([df1Car, df2Car], ignore_index=True)
dfBike = pd.concat([df1Bike, df2Bike], ignore_index=True)

df = pd.concat([dfBike, dfCar], axis=1)
df
```

```python
dfCar = pd.concat([df1Car, df2Car], ignore_index=True)
dfBike = pd.concat([df1Bike, df2Bike], ignore_index=True)

df = dfCar.merge(dfBike, how='inner')
df
```

```python
dfCar = pd.concat([df1Car, df2Car], ignore_index=True)
dfBike = pd.concat([df1Bike, df2Bike], ignore_index=True)

df = dfCar.merge(dfBike, how='left')
df
```

```python
dfCar = pd.concat([df1Car, df2Bike], ignore_index=True)
dfBike = pd.concat([df1Bike, df2Bike], ignore_index=True)

df = dfCar.merge(dfBike, how='right')
df
```
