```python
import numpy as np
import pandas as pd
```

```python
data = np.arange(15, 30).reshape(5, 3)
dfx = pd.DataFrame(data, index=['apple', 'banana', 'kiwi', 'grapes', 'mango'],
                   columns=['store1', 'store2', 'store3'])
```

```python
dfx['store4'] = np.nan
dfx.loc['watermelon'] = np.arange(15, 19)
dfx.loc['oranges'] = np.nan
dfx['store5'] = np.nan
dfx.loc['apple', 'store4'] = 20.0
dfx
```

```python
print("--- Check null mask ---")
print(dfx.isnull())
```

```python
print("\n--- Check non-null mask ---")
print(dfx.notnull())
```

```python
print("\n--- Missing values per column ---")
print(dfx.isnull().sum())
```

```python
print("\n--- Total missing values in dataframe ---")
print(dfx.isnull().sum().sum())
```

```python
print("\n--- Count of valid reported values per column ---")
print(dfx.count())
```

```python
print("--- Drop rows where ALL elements are NaN ---")
print(dfx.dropna(how='all'))
```

```python
print("\n--- Drop columns where ALL elements are NaN ---")
print(dfx.dropna(how='all', axis=1))
```

```python
print("\n--- Keep columns with at least 5 non-NaN values ---")
print(dfx.dropna(thresh=5, axis=1))
```

```python
print("--- Math behaviors with NaN ---")
print(f"Store 4 Total Sum: {dfx.store4.sum()}")
print(f"Store 4 Mean: {dfx.store4.mean()}")
```

```python
print("\n--- Fill all NaNs with 0 ---")
filledDf = dfx.fillna(0)
print(filledDf)
```

```python
print("\n--- Forward Fill (ffill) Column 4 ---")
print(dfx.store4.ffill())
```

```python
print("\n--- Backward Fill (bfill) Column 4 ---")
print(dfx.store4.bfill())
```
